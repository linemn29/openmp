```python
from transformers import AutoModelForCausalLM, AutoTokenizer

# Charger un modèle GPT pré-entraîné
model_name = "gpt2"  # Ou "EleutherAI/gpt-neo-125M" pour un modèle plus grand
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Entrée utilisateur
input_text = "Bonjour, comment puis-je t'aider aujourd'hui ?"
inputs = tokenizer.encode(input_text, return_tensors="pt")

# Générer une réponse
outputs = model.generate(inputs, max_length=50, num_return_sequences=1)
response = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(response)
```

```python
import gradio as gr

def chatbot(input_text):
    inputs = tokenizer.encode(input_text, return_tensors="pt")
    outputs = model.generate(inputs, max_length=50, num_return_sequences=1)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

interface = gr.Interface(fn=chatbot, inputs="text", outputs="text")
interface.launch()
```

```python
import os
from PyPDF2 import PdfReader
from langchain.embeddings import OpenAIEmbeddings
from langchain.vectorstores import FAISS
from langchain.chains import RetrievalQA
from langchain.chat_models import ChatOpenAI
from langchain.text_splitter import CharacterTextSplitter

# Configuration OpenAI API
os.environ["OPENAI_API_KEY"] = "your_openai_api_key"

# Fonction pour extraire le texte d'un PDF
def extract_text_from_pdf(pdf_path):
    reader = PdfReader(pdf_path)
    text = ""
    for page in reader.pages:
        text += page.extract_text()
    return text

# Charger un PDF
pdf_path = "/content/CV_Lyna_Mnasri.pdf"  # Remplacez par votre fichier PDF
pdf_text = extract_text_from_pdf(pdf_path)

# Diviser le texte pour les traitements LLM
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
text_chunks = text_splitter.split_text(pdf_text)

# Créer des embeddings pour le texte extrait
embeddings = OpenAIEmbeddings()
vector_store = FAISS.from_texts(text_chunks, embeddings)

# Configurer le modèle et le système de récupération (RAG)
retriever = vector_store.as_retriever()
qa_chain = RetrievalQA.from_chain_type(
    llm=ChatOpenAI(model="gpt-4"),
    retriever=retriever,
    return_source_documents=True,
)

# Poser une question basée sur le contenu du PDF
query = "Quels sont les points clés mentionnés dans le document ?"
response = qa_chain.run(query)

# Résultat
print("Réponse :", response)
```

```python
from fastapi import FastAPI, UploadFile, HTTPException
from langchain.document_loaders import PyPDFLoader
from langchain.embeddings import OpenAIEmbeddings
from langchain.vectorstores import Pinecone
import pinecone
import os
from dotenv import load_dotenv

# Charger les variables d'environnement
load_dotenv()

# Configurer Pinecone et OpenAI
PINECONE_API_KEY = os.getenv("PINECONE_API_KEY")
PINECONE_ENV = os.getenv("PINECONE_ENVIRONMENT")
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

pinecone.init(api_key=PINECONE_API_KEY, environment=PINECONE_ENV)
index_name = "pdf-extraction"
if index_name not in pinecone.list_indexes():
    pinecone.create_index(index_name, dimension=1536)  # Ajuste selon ton embedding

app = FastAPI()

@app.post("/upload/")
async def upload_pdf(file: UploadFile):
    # Vérifie que le fichier est un PDF
    if file.content_type != "application/pdf":
        raise HTTPException(status_code=400, detail="Le fichier doit être un PDF.")

    # Charger et traiter le fichier
    loader = PyPDFLoader(file.file)
    documents = loader.load()

    # Créer les embeddings et indexer
    embeddings = OpenAIEmbeddings(openai_api_key=OPENAI_API_KEY)
    vector_store = Pinecone.from_documents(documents, embeddings, index_name=index_name)

    return {"message": "PDF indexé avec succès", "file_name": file.filename}

@app.get("/query/")
async def query_index(query: str):
    # Récupère un outil de recherche
    vector_store = Pinecone(index_name=index_name)
    retriever = vector_store.as_retriever()

    # Utilise LangChain pour RAG
    from langchain.chains import RetrievalQA
    from langchain.chat_models import ChatOpenAI

    llm = ChatOpenAI(model="gpt-4", openai_api_key=OPENAI_API_KEY)
    qa_chain = RetrievalQA.from_chain_type(llm=llm, retriever=retriever)

    response = qa_chain.run(query)
    return {"query": query, "response": response}
```
